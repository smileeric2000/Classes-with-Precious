## Computer vision 
    is a field of artificial intelligence (AI) that enables computers and systems to derive meaningful information from digital images, videos, and other visual inputs — and to take actions or make recommendations based on that information.

#### Key Areas in Computer Vision:
- Image Classification
Assigning a label to an image (e.g., “cat”, “dog”).
Example: Detecting if an X-ray image shows signs of pneumonia.

- Object Detection
Identifying and locating objects in an image.
Example: Detecting cars and pedestrians in self-driving car systems.

- Image Segmentation
Dividing an image into multiple regions or objects.
Semantic segmentation: Label each pixel with a class.

- Instance segmentation: Also distinguishes between different instances of the same class.

- Facial Recognition
Identifying or verifying a person from an image or video.

- Optical Character Recognition (OCR)
Extracting text from images or documents.

- Image Generation and Enhancement
Super-resolution, image inpainting, style transfer, GANs (Generative Adversarial Networks).

- Pose Estimation
Detecting the human body’s orientation and position in an image.

- Action Recognition
Identifying actions in a video (e.g., running, waving).

##### Rotate

```python
(h, w) = img.shape[:2]   #This code grabs the height and width values of the image
center = (w // 2, h // 2) #This gets the centre of each part(height, width)
                          #It will be the rotation center.
```

```python
matrix = cv2.getRotationMatrix2D(center, 45, 1.0)  #Makes a 2D rotation of the image

'''
Arguments:
    center: point to rotate around
    
    45: rotation angle in degrees (positive = counterclockwise)
    
    1.0: scale factor (1.0 = keep size the same)
'''
```

```python
rotated = cv2.warpAffine(img, matrix, (w, h)) #This applies the rotation to the image
```

```python
plt.imshow(rotated)
plt.axis('off')
plt.show()
```

## Apply filters (blur, edge detection)
    Blurring – for smoothing out images.
    Edge Detection – for finding outlines and shapes.

#### Why blur?
    Removes noise.
    Smoothens transitions.
    Helps in preprocessing before detecting edges or features.

#### Blur 
    There are different types of blurring;
    Averaging blur
    Gaussian blur and
    Median blur

##### Averaging Blur
    Takes the average of surrounding pixels.

```python
blurred = cv2.blur(img, (20, 20))  #(20,20) is the kernel size. Bigger = more blur.
```

```python
plt.imshow(blurred)
plt.axis('off')
plt.show()
```

##### Gaussian Blur
    Gives more natural-looking blur by applying a weighted average.

```python
gaussian = cv2.GaussianBlur(img, (5, 5), 0)
#Last argument (0) is standard deviation. Keep it 0 to let OpenCV auto-calculate.
```

```python
plt.imshow(gaussian)
plt.axis('off')
plt.show()
```

##### Median Blur
    Great for removing salt-and-pepper noise.

```python
median = cv2.medianBlur(img, 27)
#Only one number needed for kernel size (must be odd). The higher the number the more it is blur
```

```python
plt.imshow(median)
plt.axis('off')
plt.show()
```

#### Edge Detection
    Edges = sharp changes in intensity.
    Helps detect object boundaries.    

##### Sobel Filter (Directional edge detection)
    Detects edges in x or y direction.

```python
cv2.CV_64F
```

```python
sobelx = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=5)  # Horizontal
sobely = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=5)  # Vertical
```

        Sobel operator is an edge detector that finds changes in intensity (brightness) in an image — essentially, it detects edges.
    
    It calculates the gradient (rate of change) of the image pixel values.



| Parameter          | Meaning                                                       |
|--------------------|---------------------------------------------------------------|
| img                | Input image (usually grayscale)                              |
| cv2.CV_64F         | Data type for output image (64-bit float) to capture negative gradients properly |
| 1, 0 (first call)  | dx=1, dy=0 → Take gradient in x-direction (horizontal) only |
| 0, 1 (second call) | dx=0, dy=1 → Take gradient in y-direction (vertical) only   |
| ksize=5            | Size of the Sobel kernel (5x5 filter) — bigger kernel smooths noise more |

sobelx: Highlights vertical edges because changes along x (horizontal axis) correspond to vertical boundaries.

sobely: Highlights horizontal edges because changes along y (vertical axis) correspond to horizontal boundaries.


##### Why use gradients?
    Edges are locations where pixel intensities change sharply.
    Sobel operator calculates how much and where those changes happen.



```python
plt.subplot(1,3,1)
plt.imshow(sobelx)
plt.xlabel("horizontal edge detection")
plt.xticks([])  # removes x-axis ticks
plt.yticks([]) 
plt.subplot(1,3,2)
plt.imshow(sobely)
plt.xlabel("vertical edge detection")
plt.xticks([])  
plt.yticks([]) 
plt.subplot(1,3,3)
plt.imshow(img)
plt.xticks([])  
plt.yticks([]) 
plt.xlabel("normal image")
plt.show()
```

##### Canny Edge Detection (Most popular)
    Smart edge detector using gradients and thresholds.

###### Canny Edge Detection detects edges by:

    Smoothing the image to reduce noise using a Gaussian blur.
    Finding gradients to identify areas with rapid intensity change.
    Thinning edges by keeping only local maxima (non-maximum suppression).
    Applying two thresholds to classify strong and weak edges.
    Tracking edges by connecting weak edges to strong ones and discarding others.

```python
edges = cv2.Canny(img, 100, 200)

'''
Pixels with gradient intensity > 200 are strong edges.
Pixels with gradient intensity between 100 and 200 are weak edges (kept only if connected to strong edges).
Pixels < 100 are discarded.

#100 = lower threshold
# 200 = upper threshold
# Lower value = more sensitive.
'''
```

```python
plt.figure(figsize = (5,3))
plt.imshow(edges
           )
plt.axis('off')
plt.show()
```

#### Summary

| Filter             | Use                      |

| cv2.GaussianBlur() | Natural blur, better quality |
| cv2.medianBlur()   | Removes salt-and-pepper noise |
| cv2.Sobel()        | Finds vertical/horizontal edges |
| cv2.Laplacian()    | Finds all edges          |
| cv2.Canny()        | Smart edge detection     |


## Contour Detection (Finding object shapes)

What is a contour?

    A contour is simply a curve joining all continuous points along a boundary of the same color or intensity — like outlining an object in an image.

#### Steps to detect contours (objects):
    Convert the image to grayscale
    Apply thresholding or edge detection
    Use cv2.findContours() to find the contours
    Draw them with cv2.drawContours()

###### convert image to gray

```python
#convert image to grey
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
#Use Canny for better contour detection
edges = cv2.Canny(gray, 50, 150)
```

```python
#Find contours
contours, hierarchy = cv2.findContours(edges, cv2.RETR_EXTERNAL , cv2.CHAIN_APPROX_SIMPLE)

'''
Describing the realized variables(contours, heirarchy)
*** contours: A list of NumPy arrays, where each array is a contour.
Each contour is just a set of points outlining a shape.
*** hierarchy: A NumPy array that shows the relationship between 
contours (like parent-child).
Especially useful when using cv2.RETR_TREE or cv2.RETR_CCOMP.
Each contour has 4 values: [Next, Previous, First_Child, Parent] index
of the contour 

HYPERPARAMETERS: 
1st arg (image_data<must be binary image i.e black & white>)
2nd arg (mode; defines how many contours to retrieve and their hierarchy)
'cv2.RETR_EXTERNAL'	Retrieves only outer contours (ignores holes or nested shapes i.e contours<objects> within other contours<objects>)
However if you use cv2.RETR_TREE instead, a full hierarchy of all contours, including:
    Outer shapes
    Holes or inner contours
    Nested contours (even third or fourth-level nesting!)
    and it gives you relationship data, in hierarchy, like which contour 
    is the parent and which is child
3rd arg (method;how to approximate contours points)
'''
```

##### Print a list of all distinct objects(contours) in the image and their sizes

```python
for cnt in contours:
    area = cv2.contourArea(cnt)
    print("Area:", area)
```

```python
contour_img = cv2.drawContours(img, contours, -1, (0, 255, 0), 2)
```

#### Print image after contours (objects) have been detected

```python
# Draw contours
plt.figure(figsize = (5,3))
plt.imshow(contour_img)
plt.axis('off')
plt.show()
```
